# 05 — IQM architectures and native compilation

**Level:** intermediate. **Environment:** `iqm-client` / `qiskit-iqm` fake backend; CPU-only and credential-free. **Time:** seconds. **Result:** architecture-constrained transpilation and noisy sample outcomes.

IQM's fake backend models a *representative* error profile; it is **not** a connection to an IQM machine and must not be used as actual device calibration evidence.

```python
from qiskit import QuantumCircuit, transpile
from iqm.qiskit_iqm import IQMFakeAdonis

backend = IQMFakeAdonis()
circuit = QuantumCircuit(2)
circuit.h(0)
circuit.cx(0,1)
circuit.measure_all()
compiled = transpile(circuit, backend=backend)
job = backend.run(compiled, shots=512)
counts = job.result().get_counts()
assert sum(counts.values()) == 512
print("Fake Adonis operations:", compiled.count_ops())
print("Noisy simulated counts:", counts)
```
