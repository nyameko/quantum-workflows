# 04 — Cirq, qsim and NVIDIA cuQuantum: know the boundaries

**Level:** intermediate. **Local environment:** Cirq + optional qsimcirq on CPU. **HPC:** qsim CUDA/cuStateVec and cuQuantum backends require dedicated validated GPU images and Slurm allocations. **Time:** seconds locally. **Outputs:** reference sampled Bell result.

Cirq is a circuit framework; qsim is a simulator; cuStateVec/cuTensorNet are acceleration libraries. Do not label these a physical QPU.

```python
import cirq
q0, q1 = cirq.LineQubit.range(2)
circuit = cirq.Circuit(cirq.H(q0), cirq.CNOT(q0,q1), cirq.measure(q0,q1,key="z"))
result = cirq.Simulator(seed=42).run(circuit, repetitions=1024)
hist = result.histogram(key="z")
assert set(hist) <= {0,3}, hist
print("Cirq CPU Bell counts:", dict(hist))
```

```python
try:
    import qsimcirq
except ImportError:
    print("qsimcirq is not installed in this CPU kernel; use the separate qsim environment.")
else:
    qsim_result = qsimcirq.QSimSimulator().run(circuit, repetitions=1024)
    assert set(qsim_result.histogram(key="z")) <= {0,3}
    print("qsim CPU Bell counts:", dict(qsim_result.histogram(key="z")))
```
