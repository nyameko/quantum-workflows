# 06 — Pasqal/Pulser: neutral-atom analogue emulation

**Level:** intermediate. **Environment:** Pulser + `pulser-simulation` on CPU. **Optional scale-up:** validated `emu-sv` / `emu-mps` backend on an approved GPU allocation. **Time:** seconds to minutes. **Result:** a pulse sequence and numerical emulator output.

Neutral-atom analogue Hamiltonian evolution is not interchangeable with gate-circuit simulation; state the device geometry and approximations.

```python
import numpy as np
from pulser import Register, Sequence, Pulse
from pulser.devices import MockDevice
from pulser_simulation import QutipEmulator

register = Register.from_coordinates([(0,0), (6,0)], prefix="q")
seq = Sequence(register, MockDevice)
seq.declare_channel("rydberg", "rydberg_global")
seq.add(Pulse.ConstantPulse(200, 2*np.pi, 0.0, 0.0), "rydberg")
emu = QutipEmulator.from_sequence(seq)
result = emu.run()
print("Rydberg pulse sequence:", seq)
print("CPU emulation result:", type(result).__name__)
```
