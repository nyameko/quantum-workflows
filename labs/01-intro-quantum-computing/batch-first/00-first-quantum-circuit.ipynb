# 00 — First quantum circuit: Bell pair on CPU

**Principle:** scientific preparation is cheap; expensive execution is scheduled.

**Level:** beginner. **Environment:** Qiskit + CPU Qiskit Aer, in the ordinary Kubernetes JupyterHub workbench. **Run time:** seconds. **Credentials:** none. **Output:** Bell-state counts. **Cleanup:** none.

Outcomes: apply H/CX, measure a Bell state, interpret shot statistics and distinguish CPU simulator from GPU scheduler and remote physical QPU. See also the existing `../Intro_QC_Demo.ipynb` and `../02-bell-cpu-gpu-qpu.ipynb`; this is a shorter on-ramp, not a replacement.

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator

circuit = QuantumCircuit(2, 2)
circuit.h(0)
circuit.cx(0, 1)
circuit.measure([0, 1], [0, 1])
sim = AerSimulator(method="statevector", device="CPU")
result = sim.run(transpile(circuit, sim), shots=1024, seed_simulator=42).result()
counts = result.get_counts()
assert result.success and sum(counts.values()) == 1024
assert set(counts) <= {"00", "11"}, counts
print("Bell counts:", counts)
```

## Interpret

An ideal Z-basis Bell measurement yields only `00` and `11` outcomes. It does not **alone** certify entanglement. In notebook 01 you will submit a larger GPU experiment as a durable batch job, without allocating a GPU to this workbench.
