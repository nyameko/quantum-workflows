**Level:** intermediate. **Local environment:** Qiskit Aer CPU. **Remote:** institutionally approved physical QPU provider using its *native* SDK. **Time/cost:** provider queue, shots and credits vary; no background submission. **Outputs:** local reference plus approved provider execution/provenance (future acceptance gate).

A development framework, a simulator backend and a physical QPU are not the same resource. **No physical hardware submission is performed by this notebook.** Only execute on an approved provider after programme entitlement, shot budget, exact device identity and explicit consent are verified by Quantum Platform.

```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
circuit = QuantumCircuit(2, 2)
circuit.h(0)
circuit.cx(0,1)
circuit.measure([0,1],[0,1])
sim = AerSimulator(device="CPU")
local = sim.run(transpile(circuit, sim), shots=256, seed_simulator=42).result()
counts = local.get_counts()
assert set(counts) <= {"00","11"} and sum(counts.values()) == 256
print("Credential-free local reference:", counts)
```

```python
approval = {
    "programme_authorized": False,
    "provider_backend_id": None,
    "allocated_shots": 0,
    "user_approved_cost": False,
    "credentials_secret_ref": None,
}
required = (approval["programme_authorized"] and approval["provider_backend_id"]
            and approval["allocated_shots"] > 0 and approval["user_approved_cost"]
            and approval["credentials_secret_ref"])
print("Physical QPU submission enabled?", bool(required))
# Intentionally no provider.run() call. Implement only after reviewed access-policy integration.
```
